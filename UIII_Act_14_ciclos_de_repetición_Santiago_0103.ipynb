## Problema:
#### Algoritmo que convierta un número decimal a hexadecimal (0-9, A-F) usando bucle y módulo.
![link text](https://raw.githubusercontent.com/SantiagoMedrano128/UIII-Act-14-ciclos-de-repetici-n-/refs/heads/main/infoact14.png)

```python
# --- ALGORITMO DE CONVERSIÓN DECIMAL A HEXADECIMAL ---

# Solicita al usuario un número entero decimal
numero = int(input("Ingresa un número decimal positivo: "))

# Variable para construir la cadena de texto con el resultado
resultado_hex = ""

# Caso especial: si el número es 0, el resultado es directamente "0"
if numero == 0:
    resultado_hex = "0"
else:
    # Ciclo que se repite mientras el número sea mayor a cero [3, 4]
    while numero > 0:
        # Obtenemos el residuo de la división entre 16 (base hexadecimal) [1, 5]
        residuo = numero % 16

        # Convertimos el residuo numérico a su carácter hexadecimal [2, 6]
        if residuo == 10:
            caracter = "A"
        elif residuo == 11:
            caracter = "B"
        elif residuo == 12:
            caracter = "C"
        elif residuo == 13:
            caracter = "D"
        elif residuo == 14:
            caracter = "E"
        elif residuo == 15:
            caracter = "F"
        elif residuo == 16:
            caracter = "G"
        elif residuo == 17:
            caracter = "H"
        elif residuo == 18:
            caracter = "I"
        elif residuo == 19:
            caracter = "J"
        elif residuo == 20:
            caracter = "K"
        elif residuo == 21:
            caracter = "L"
        elif residuo == 22:
            caracter = "M"
        elif residuo == 23:
            caracter = "N"
        elif residuo == 24:
            caracter = "O"
        elif residuo == 25:
            caracter = "P"
        elif residuo == 26:
            caracter = "Q"
        elif residuo == 27:
            caracter = "R"
        elif residuo == 28:
            caracter = "S"
        elif residuo == 29:
            caracter = "T"
        elif residuo == 30:
            caracter = "U"
        elif residuo == 31:
            caracter = "V"
        elif residuo == 32:
            caracter = "W"
        elif residuo == 33:
            caracter = "X"
        elif residuo == 34:
            caracter = "Y"
        elif residuo == 35:
            caracter = "Z"
        else:
            # Si el residuo es 0-9, se convierte simplemente a texto [7]
            caracter = str(residuo)

        # Concatenamos el carácter encontrado al INICIO de la cadena acumulada [8, 9]
        resultado_hex = caracter + resultado_hex

        # Actualizamos el número usando división entera para la siguiente iteración [1, 10]
        numero = numero // 16

# Muestra el resultado final en pantalla [11, 12]
print("El número en hexadecimal es:", resultado_hex)
```
